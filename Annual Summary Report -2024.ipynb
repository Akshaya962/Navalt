```python
import numpy as np
import pandas as pd
import warnings
import matplotlib.pyplot as plt
import requests
import io

warnings.filterwarnings("ignore")
```

```python
MSC_noon_2024=pd.read_csv("msc_2024.csv")
                                                                                                                                    
```

```python
pd.set_option('display.max_columns', None)  
MSC_noon_2024
```

```python
MSC_noon_2024['Date'] = pd.to_datetime(MSC_noon_2024['corrected_date']).apply(lambda x:x.strftime("%d"))
MSC_noon_2024['Month'] = pd.to_datetime(MSC_noon_2024['corrected_date']).apply(lambda x:x.strftime("%m"))
MSC_noon_2024['Year'] = pd.to_datetime(MSC_noon_2024['corrected_date']).apply(lambda x:x.strftime("%Y"))
MSC_noon_2024['Month_name'] = pd.to_datetime(MSC_noon_2024['corrected_date']).apply(lambda x:x.strftime("%B"))
MSC_noon_2024['date'] = pd.to_datetime(MSC_noon_2024["corrected_date"]).dt.date
MSC_noon_2024.dropna(subset=['imo'])
MSC_noon_2024_sort=MSC_noon_2024.sort_values(by="corrected_date")
#Total HS,LS,MGO and Total Fuel
MSC_noon_2024_sort['Total_Calculated_HS']= MSC_noon_2024_sort[['fuel_me_rsdl_hs' , 'fuel_aux_rsdl_hs' , 'fuel_boiler_rsdl_hs']].sum(axis=1)
MSC_noon_2024_sort['Total_Calculated_LS'] = MSC_noon_2024_sort[['fuel_me_rsdl_vls' , 'fuel_me_dstlt_vls' , 'fuel_me_tnktnr_dstlt_vls' ,'fuel_aux_rsdl_vls' ,  'fuel_aux_dstlt_vls' , 'fuel_aux_tnktnr_dstlt_vls','fuel_boiler_rsdl_vls','fuel_boiler_dstlt_vls','fuel_boiler_tnktnr_dstlt_vls']].fillna(0).sum(axis=1)
MSC_noon_2024_sort['Total_Calculated_ULS'] = MSC_noon_2024_sort[['fuel_me_rsdl_uls' , 'fuel_aux_rsdl_uls' , 'fuel_boiler_rsdl_uls' ,'fuel_me_dstlt_uls' ,'fuel_aux_dstlt_uls' , 'fuel_boiler_dstlt_uls']].fillna(0).sum(axis=1)
MSC_noon_2024_sort['Total_Calculated_Fuel'] = MSC_noon_2024_sort[['Total_Calculated_HS' ,'Total_Calculated_LS' , 'Total_Calculated_ULS']].fillna(0).sum(axis=1)
MSC_noon_2024_sort['Total_Calculated_LNG'] = MSC_noon_2024_sort[['fuel_m_e_lng' ,'fuel_boiler_lng' , 'fuel_aux_lng']].fillna(0).sum(axis=1)

#HS_CO2,LSCO2,MGO_CO2 and Total CO2
MSC_noon_2024_sort['HS_CO2'] = 3.114 * MSC_noon_2024_sort["Total_Calculated_HS"]
MSC_noon_2024_sort['LS_CO2'] = 3.151 * MSC_noon_2024_sort["Total_Calculated_LS"]
MSC_noon_2024_sort['ULS_CO2'] = 3.206 * MSC_noon_2024_sort["Total_Calculated_ULS"]
MSC_noon_2024_sort['Total_CO2'] = MSC_noon_2024_sort[['HS_CO2' ,'LS_CO2' ,'ULS_CO2']].fillna(0).sum(axis=1)

MSC_noon_2024_sort["fuel_m_e_calculated"]=MSC_noon_2024_sort[["fuel_me_rsdl_hs","fuel_me_rsdl_vls","fuel_me_dstlt_vls",'fuel_me_rsdl_uls','fuel_me_rsdl_uls','fuel_me_tnktnr_dstlt_vls']].fillna(0).sum(axis=1)
MSC_noon_2024_sort["fuel_aux_calculated"]=MSC_noon_2024_sort[["fuel_aux_rsdl_vls","fuel_aux_rsdl_hs","fuel_aux_dstlt_vls",'fuel_aux_rsdl_uls','fuel_aux_dstlt_uls','fuel_aux_tnktnr_dstlt_vls']].fillna(0).sum(axis=1)
MSC_noon_2024_sort["fuel_boiler_calculated"]=MSC_noon_2024_sort[["fuel_boiler_rsdl_hs","fuel_boiler_rsdl_vls","fuel_boiler_dstlt_vls",'fuel_boiler_rsdl_uls','fuel_boiler_tnktnr_dstlt_vls','fuel_boiler_dstlt_uls']].fillna(0).sum(axis=1)
MSC_noon_2024_sort["Total_fuel_calculated"]=MSC_noon_2024_sort[["fuel_m_e_calculated","fuel_aux_calculated","fuel_boiler_calculated"]].fillna(0).sum(axis=1)
```

```python

MSC_noon_2024_str = MSC_noon_2024_sort[MSC_noon_2024_sort['voyage'].apply(lambda x: isinstance(x, str))]
MSC_noon_2024_str['voyage'] = MSC_noon_2024_str['voyage'].str.lstrip('0').str.rstrip('.0')

# Filter by Float Type and Convert to String
MSC_noon_2024_float = MSC_noon_2024_sort[MSC_noon_2024_sort['voyage'].apply(lambda x: isinstance(x, float))]
MSC_noon_2024_float['voyage'] = pd.to_numeric(MSC_noon_2024_float['voyage'], errors='coerce').fillna(0).astype(int).astype(str)

# Filter by Integer Type and Convert to String
MSC_noon_2024_int = MSC_noon_2024_sort[MSC_noon_2024_sort['voyage'].apply(lambda x: isinstance(x, int))]
MSC_noon_2024_int['voyage'] = MSC_noon_2024_int['voyage'].astype(int).astype(str)

# Concatenate DataFrames and Sort
MSC_noon_2024_sort_new = pd.concat([MSC_noon_2024_str, MSC_noon_2024_float, MSC_noon_2024_int])
MSC_noon_2024_sort_new = MSC_noon_2024_sort_new.sort_values(by=["corrected_date"])
```

```python
cargo_vessel_mean_2024 = MSC_noon_2024_sort_new.groupby(["imo", "voyage"])[["cargo_total"]].mean().reset_index()
Total_cargo_2024 = cargo_vessel_mean_2024[["cargo_total"]].sum().reset_index()
Total_cargo_2024 = round(Total_cargo_2024[0] / (10 ** 6), 2)
Total_cargo_2024.reset_index()
```

# 3.  Changing unit of Fuel

```python

# Fuel calculations
fuel_hs_2024 = MSC_noon_2024_sort.groupby(["imo"])[["Total_Calculated_HS"]].sum().reset_index()
Total_fuel_hs_2024 = fuel_hs_2024[["Total_Calculated_HS"]].sum().reset_index()
Total_fuel_hs_2024 = Total_fuel_hs_2024[0] / (10 ** 6)
Total_fuel_hs_2024.reset_index()

fuel_ls_2024 = MSC_noon_2024_sort.groupby(["imo"])[["Total_Calculated_LS"]].sum().reset_index()
Total_fuel_ls_2024 = fuel_ls_2024[["Total_Calculated_LS"]].sum().reset_index()
Total_fuel_ls_2024 = Total_fuel_ls_2024[0] / (10 ** 6)
Total_fuel_ls_2024.reset_index()

fuel_uls_2024 = MSC_noon_2024_sort.groupby(["imo"])[["Total_Calculated_ULS"]].sum().reset_index()
Total_fuel_uls_2024 = fuel_uls_2024[["Total_Calculated_ULS"]].sum().reset_index()
Total_fuel_uls_2024 = Total_fuel_uls_2024[0] / (10 ** 6)
Total_fuel_uls_2024.reset_index()

fuel_lng_2024 = MSC_noon_2024_sort.groupby(["imo"])[["Total_Calculated_LNG"]].sum().reset_index()
Total_fuel_lng_2024 = fuel_lng_2024[["Total_Calculated_LNG"]].sum().reset_index()
Total_fuel_lng_2024 = Total_fuel_lng_2024[0] / (10 ** 6)
Total_fuel_lng_2024.reset_index()

fuel_me_2024 = MSC_noon_2024_sort.groupby(["imo"])[["fuel_m_e_calculated"]].sum().reset_index()
Total_fuel_me_2024 = fuel_me_2024[["fuel_m_e_calculated"]].sum().reset_index()
Total_fuel_me_2024 = Total_fuel_me_2024[0] / (10 ** 6)
Total_fuel_me_2024.reset_index()

fuel_aux_2024 = MSC_noon_2024_sort.groupby(["imo"])[["fuel_aux_calculated"]].sum().reset_index()
Total_fuel_aux_2024 = fuel_aux_2024[["fuel_aux_calculated"]].sum().reset_index()
Total_fuel_aux_2024 = Total_fuel_aux_2024[0] / (10 ** 6)
Total_fuel_aux_2024.reset_index()

fuel_boiler_2024 = MSC_noon_2024_sort.groupby(["imo"])[["fuel_boiler_calculated"]].sum().reset_index()
Total_fuel_boiler_2024 = fuel_boiler_2024[["fuel_boiler_calculated"]].sum().reset_index()
Total_fuel_boiler_2024 = Total_fuel_boiler_2024[0] / (10 ** 6)
Total_fuel_boiler_2024.reset_index()

fuel_2024 = MSC_noon_2024_sort.groupby(["imo"])[["Total_fuel_calculated"]].sum().reset_index()
Total_fuel_2024 = fuel_2024[["Total_fuel_calculated"]].sum().reset_index()
Total_fuel_2024 = Total_fuel_2024[0] / (10 ** 6)
Total_fuel_2024.reset_index()

# CO2 calculations
fuel_CO2_2024 = MSC_noon_2024_sort.groupby(["imo"])[["Total_CO2"]].sum().reset_index()
Total_CO2_2024 = fuel_CO2_2024[["Total_CO2"]].sum().reset_index()
Total_CO2_2024 =Total_CO2_2024[0] / (10 ** 6)
Total_CO2_2024.reset_index()
```
