```python
sail_df = pd.DataFrame(data['data']['SAIL']['table_1'])
port_df = pd.DataFrame(data['data']['PORT']['table_1'] + data['data']['PORT']['table_2'])
```

```python
# Concatenate sail and port data into a single DataFrame
df = pd.concat([sail_df, port_df], ignore_index=True)
# Define parameters
start_date = '2017-07-09'
end_date = '2024-07-07'
sail_or_port_data = 'SAIL'  # or 'PORT' or 'BOTH'
no_of_profiles = 3  # Set the number of profiles/clusters you want

# Select the relevant DataFrame based on the status
if sail_or_port_data == 'SAIL':
    df = sail_df.copy()
elif sail_or_port_data == 'PORT':
    df = port_df.copy()
else:
    raise ValueError("Invalid value for sail_or_port_data. Choose 'SAIL' or 'PORT'.")




if 'sea_state' in df.columns:
    df['new_ss'] = 10 - df['sea_state']
```

```python
sail_df
```

```python
if sail_or_port_data == 'SAIL':
    df['new_ss'] = 10 - df['sea_state']
    clusterdf = df[['speed', 'mean_draft', 'new_ss', 'me_actual_steaming_time', 'ae_t_steaming', 'aux_running', 'blr_running']].copy()
elif sail_or_port_data == 'PORT':
    clusterdf = df[['mean_draft', 'ae_t_steaming', 'aux_running', 'blr_running', 'activity_time']].copy()

# Clean the data
numeric_df = clusterdf.select_dtypes(include=[np.float64, np.int64])
numeric_df.fillna(numeric_df.mean(), inplace=True)
numeric_df.replace([float('inf'), float('-inf')], 1e10, inplace=True)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df_scaled = scaler.fit_transform(numeric_df)
```

```python
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
import pandas as pd
from scipy.cluster.hierarchy import fcluster, linkage

# Apply t-SNE
tsne = TSNE(n_components=2, perplexity=30, n_iter=300)
df_tsne = tsne.fit_transform(df_scaled)

# Convert to a DataFrame
df_tsne = pd.DataFrame(df_tsne, columns=['TSNE1', 'TSNE2'])

# Visualize the t-SNE result
plt.figure(figsize=(10, 6))
plt.scatter(df_tsne['TSNE1'], df_tsne['TSNE2'])
plt.title('t-SNE Visualization of Vessel Operational Data')
plt.xlabel('TSNE1')
plt.ylabel('TSNE2')
plt.show()

# Perform hierarchical clustering (without dendrogram)
linked = linkage(df_tsne, method='ward')

# Cut the linkage to form clusters
cluster_labels = fcluster(linked, no_of_profiles, criterion='maxclust')

# Add the cluster labels to the original t-SNE DataFrame
df_tsne['Cluster'] = cluster_labels

# Plot the clusters
plt.figure(figsize=(10, 6))
for cluster in range(1, no_of_profiles + 1):
    subset = df_tsne[df_tsne['Cluster'] == cluster]
    plt.scatter(subset['TSNE1'], subset['TSNE2'], label=f'Cluster {cluster}')

plt.title('t-SNE Clusters of Vessel Operational Data')
plt.xlabel('TSNE1')
plt.ylabel('TSNE2')
plt.legend()
plt.show()
```

```python
clusterdf['Cluster'] = df_tsne['Cluster'].values

if sail_or_port_data == 'SAIL':
    # Calculate sail-specific profiles
    clusterdf['sea_state'] = 10 - clusterdf['new_ss']
    clusterdf['sea_state'] = clusterdf['sea_state'].apply(lambda x: int(x))
    clusterdf.drop('new_ss', axis=1, inplace=True)
    
    cluster_profile = clusterdf.groupby(['Cluster']).mean().reset_index().round(2)
    me_time = (clusterdf.groupby('Cluster')['me_actual_steaming_time'].sum() / 24).reset_index().round(2)
    sail_profile_data = pd.merge(cluster_profile, me_time, on='Cluster', how='inner')
    sail_profile_data.rename(columns={
        'me_actual_steaming_time_x': 'me_actual_steaming_time_mean',
        'me_actual_steaming_time_y': 'me_actual_steaming_time_days'
    }, inplace=True)
    sail_profile_data['sea_state'] = sail_profile_data['sea_state'].apply(lambda x: round(x))
    sail_profile_data['aux_running'] = sail_profile_data['aux_running'].apply(lambda x: round(x))
    sail_profile_data['blr_running'] = sail_profile_data['blr_running'].apply(lambda x: round(x))
    print("SAIL Cluster Data:")
    print(clusterdf)
    print("SAIL Profile Data:")
    print(sail_profile_data)

elif sail_or_port_data == 'PORT':
    # Calculate port-specific profiles
    cluster_profile = clusterdf.groupby(['Cluster']).mean().reset_index().round(2)
    activity_time = (clusterdf.groupby('Cluster')['activity_time'].sum() / 24).reset_index().round(2)
    port_profile_data = pd.merge(cluster_profile, activity_time, on='Cluster', how='inner')
    port_profile_data.rename(columns={
        'activity_time_x': 'activity_time_mean',
        'activity_time_y': 'activity_time_days'
    }, inplace=True)
    port_profile_data['sea_state'] = port_profile_data['sea_state'].apply(lambda x: round(x))
    port_profile_data['aux_running'] = port_profile_data['aux_running'].apply(lambda x: round(x))
    port_profile_data['blr_running'] = port_profile_data['blr_running'].apply(lambda x: round(x))
    print("PORT Cluster Data:")
    print(clusterdf)
    print("PORT Profile Data:")
    print(port_profile_data)
```
