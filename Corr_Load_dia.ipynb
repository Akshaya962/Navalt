```python
import plotly.graph_objs as go
import pandas as pd
import numpy as np


imo_lst = list(s['data']['data'].keys())
required_curve_columns = ['shaft_speed', 'pover_load', 'p_continuous', 'max_speed', 'prop_curve_s', 'prop_curve_l', 'prop_curve_bp']
legend_labels = []


plot_divs = []


# Iterate through each load diagram dynamically
for ld_key, imos in s['data']['LoadDiagramData']['plot_points_imos_list'].items():
    fig = go.Figure()

    # Plot load diagram curves for each IMO
    for imo in imos:
        if imo in imo_lst:
            data_curve = pd.DataFrame(s['data']['LoadDiagramData']['chart_data'][ld_key])
            
            if all(col in data_curve.columns for col in required_curve_columns):
                data_curve = data_curve[required_curve_columns].astype(float)
                data_curve = data_curve.sort_values(by='shaft_speed')
                data_curve = data_curve.replace([''], np.nan)
                data_curve = data_curve.dropna(how='all')
                
                    # Plot the load diagram curves
                fig.add_trace(go.Scatter(x=data_curve['shaft_speed'], y=data_curve['pover_load'], mode='lines', name='Power Over Load', line=dict(color='red')))
                fig.add_trace(go.Scatter(x=data_curve['shaft_speed'], y=data_curve['p_continuous'], mode='lines' , name='Continuous Power', line=dict(color='blue')))
                fig.add_trace(go.Scatter(x=data_curve['max_speed'], y=data_curve['pover_load'], mode='lines' , name='Max Speed', line=dict(color='lightblue')))
                fig.add_trace(go.Scatter(x=data_curve['shaft_speed'], y=data_curve['prop_curve_s'], mode='lines', name='Prop Curve S', line=dict(color='green')))
                fig.add_trace(go.Scatter(x=data_curve['shaft_speed'], y=data_curve['prop_curve_l'], mode='lines', name='Prop Curve L', line=dict(color='violet')))
                fig.add_trace(go.Scatter(x=data_curve['shaft_speed'], y=data_curve['prop_curve_bp'], mode='lines', name='Prop Curve BP', line=dict(color='dark green')))
                
#                     fig.add_trace(go.Scatter(x=data_curve['shaft_speed'], y=data_curve['pover_load'], mode='lines'))
#                     fig.add_trace(go.Scatter(x=data_curve['shaft_speed'], y=data_curve['p_continuous'], mode='lines' ))
#                     fig.add_trace(go.Scatter(x=data_curve['max_speed'], y=data_curve['pover_load'], mode='lines'))
#                     fig.add_trace(go.Scatter(x=data_curve['shaft_speed'], y=data_curve['prop_curve_s'], mode='lines'))
#                     fig.add_trace(go.Scatter(x=data_curve['shaft_speed'], y=data_curve['prop_curve_l'], mode='lines'))
#                     fig.add_trace(go.Scatter(x=data_curve['shaft_speed'], y=data_curve['prop_curve_bp'], mode='lines'))

    # Plot scatter data for each IMO
    if types != 'both':
        if method == 'class':
            for imo in imos:
                if imo in imo_lst:
                    data_scatter_imo = pd.DataFrame(s["data"]['data'][imo])
                    if not data_scatter_imo.empty and 'vessel_name' in data_scatter_imo.columns:
                        data_scatter_imo = data_scatter_imo[['type', 'vessel_name', 'imo', 'rpm', 'power_kw', 'me_load', 'rpm_loaddia']]
                        for label in data_scatter_imo['vessel_name'].unique():
                            color = color_lst.get(imo, color_lst.get(label, '#000000'))  # Default to black if no color found
                        
                        if label not in legend_labels:
                            fig.add_trace(go.Scatter(x=data_scatter_imo['rpm_loaddia'], y=data_scatter_imo['me_load'], mode='markers', marker=dict(color=color), name=label))
                            legend_labels.append(label)
                        else:
                            fig.add_trace(go.Scatter(x=data_scatter_imo['rpm_loaddia'], y=data_scatter_imo['me_load'], mode='markers', marker=dict(color=color), showlegend=False))

                        # Handle new curve point plotting
                        new_curve = s['data']['LoadDiagramData'].get('chart_data', {}).get(ld_key, [])
                        if new_curve:
                            x_value = new_curve[0].get('x', None)
                            y_value = new_curve[0].get('y', None)
                            if x_value is not None and y_value is not None:
                                fig.add_trace(go.Scatter(x=[x_value], y=[y_value], mode='markers', marker=dict(color='yellow', size=10), name=f'New RPM and Power {ld_key}'))
    elif types == 'both':  # Data comparison
        if method == 'class':
            for imo in imos:
                if imo in imo_lst:
                    data_scatter_imo = pd.DataFrame(s["data"]['data'][imo])
                    if not data_scatter_imo.empty and 'vessel_name' in data_scatter_imo.columns:
                        data_scatter_imo = data_scatter_imo[['type', 'vessel_name', 'imo', 'rpm', 'power_kw', 'me_load', 'rpm_loaddia']]
                        data_scatter_vrs = data_scatter_imo[data_scatter_imo['type'] == 'VRS Data']
                        data_scatter_ada = data_scatter_imo[data_scatter_imo['type'] == 'ADA Data']
                        
                        if f'{ld_key} VRS Data' not in legend_labels:
                            fig.add_trace(go.Scatter(x=data_scatter_vrs['rpm_loaddia'], y=data_scatter_vrs['me_load'], mode='markers', marker=dict(color='green'), name='VRS Data'))
                            legend_labels.append('VRS Data')
                        else:
                            fig.add_trace(go.Scatter(x=data_scatter_vrs['rpm_loaddia'], y=data_scatter_vrs['me_load'], mode='markers', marker=dict(color='green')))

                        if f'{ld_key} ADA Data' not in legend_labels:
                            fig.add_trace(go.Scatter(x=data_scatter_ada['rpm_loaddia'], y=data_scatter_ada['me_load'], mode='markers', marker=dict(color='red'), name='ADA Data'))
                            legend_labels.append('ADA Data')
                        else:
                            fig.add_trace(go.Scatter(x=data_scatter_ada['rpm_loaddia'], y=data_scatter_ada['me_load'], mode='markers', marker=dict(color='red')))

    # Handle new curve point plotting
    new_curve = s['data']['LoadDiagramData'].get('chart_data', {}).get(ld_key, [])
    if new_curve:
        x_value = new_curve[0].get('x', None)
        y_value = new_curve[0].get('y', None)
        if x_value is not None and y_value is not None:
            fig.add_trace(go.Scatter(x=[x_value], y=[y_value], mode='markers', marker=dict(color='yellow', size=10), name=f'New RPM and Power {ld_key}'))

        fig.update_layout( xaxis_title='RPM(%)', yaxis_title='Power (kW%)', legend_title='Legend', height= 650, width=1100)
        plot_div = pyo.plot(fig, include_plotlyjs=True, output_type='div', config={'displayModeBar': False})
        plot_divs.append(plot_div)
```
