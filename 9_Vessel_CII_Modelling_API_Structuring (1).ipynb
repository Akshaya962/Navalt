```python
def run_xgboost(X, y):
    import pandas as pd
    import numpy as np
    from sklearn.model_selection import train_test_split, GridSearchCV
    from sklearn.metrics import mean_squared_error
    from xgboost import XGBRegressor
    from sklearn.preprocessing import StandardScaler
    from sklearn.metrics import mean_squared_error
    from sklearn.metrics import mean_squared_error, r2_score


    # Split the data into training and testing sets for Deep Learning
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)



    # Initialize and train the XGBoost model
    xgb = XGBRegressor()
    xgb.fit(X_train, y_train)

    # Predictions
    y_pred = xgb.predict(X_test)

    # Evaluation
    mse = mean_squared_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)
    print(f'Mean Squared Error: {mse}')
    print(f'R-squared: {r2}') 
    return xgb
```

```python
def cii_rating(ratio):
    d1 =  0.83
    d2 =  0.94
    d3 =  1.07
    d4 =  1.19

    if ratio  <= d1:
        return "A"         
    elif (ratio > d1) & (ratio <= d2):
        return "B"
    elif (ratio > d2) & (ratio <= d3):
        return "C"
    elif (ratio > d3) & (ratio <= d4):
        return "D"
    elif (ratio > d4) :
        return "E" 
```

```python
import pandas as pd
import plotly.graph_objects as go

def find_cii_prediction(subset_port_data, subset_sail_data, no_of_days, speed_reduction):
    predicted_co2_port = {}
    for i in subset_port_data['Cluster'].unique():
        df = subset_port_data[subset_port_data['Cluster'] == i][['mean_draft', 'ae_t_steaming', 'aux_running', 'blr_running',
                                                                 'activity_time']]
        port_co2_predicted = port_co2_model.predict(df).sum()
        predicted_co2_port['Port_CO2_' + str(i)] = round(port_co2_predicted, 2)

    predicted_co2_sail = {}
    predicted_distance_sail = {}
    
    for i in subset_sail_data['Cluster'].unique():
        df = subset_sail_data[subset_sail_data['Cluster'] == i][['speed', 'mean_draft', 'new_ss',
                                                                 'me_actual_steaming_time',
                                                                 'ae_t_steaming', 'aux_running', 'blr_running']]
        # Apply speed reduction
        df['speed'] = df['speed'] - df['speed'] * speed_reduction / 100

        # Predict CO2 emissions and distances for sailing
        sail_co2_predicted = sail_co2_model.predict(df).sum()
        sail_distance_predicted = sail_distance_model.predict(df).sum()
        
        predicted_co2_sail['Sail_CO2_' + str(i)] = round(sail_co2_predicted, 2)
        predicted_distance_sail['Sail_Distance_' + str(i)] = round(sail_distance_predicted, 2)
        
    # Combine results
    finaldata = []
    for a, b, c, d in zip(predicted_co2_sail.keys(), predicted_co2_sail.values(), predicted_distance_sail.keys(), predicted_distance_sail.values()):
        for p, q in zip(predicted_co2_port.keys(), predicted_co2_port.values()):
            finaldata.append((a, b, p, q, c, d))
            
    global finaldf
            
    finaldf = pd.DataFrame(finaldata, columns=['Sail_CO2', 'Sail_CO2_value', 'Port_CO2', 'Port_CO2_Value', 
                                               'Sail_Distance', 'Sail_Distance_Value']) 
    finaldf[['Sail_CO2_value', 'Port_CO2_Value', 'Sail_Distance_Value']] = finaldf[['Sail_CO2_value', 'Port_CO2_Value', 'Sail_Distance_Value']].astype(float).round(2)
    
    finaldf['deadweight'] = 39418
    finaldf['Sail_CO2'] = 9498.97
    finaldf['Port_CO2'] = 1235.05
    finaldf['distance'] = 29138.7
    finaldf['required_cii_2024'] = 10.44
    finaldf['attained_cii'] = (finaldf['Sail_CO2_value'] + finaldf['Sail_CO2'] + finaldf['Port_CO2'] + finaldf['Port_CO2_Value']) * 1000000 / (finaldf['deadweight'] * (finaldf['Sail_Distance_Value'] + finaldf['distance']))
    finaldf['A_R_Ratio'] = round((finaldf['attained_cii'] / finaldf['required_cii_2024']), 2) 
    finaldf['cii_rating'] = finaldf['A_R_Ratio'].apply(cii_rating)
    finaldf['avg_speed'] = round(finaldf['Sail_Distance_Value'] / (no_of_days * 24), 2)
    
    return finaldf

def find_cii_for_speed_reductions(subset_port_data, subset_sail_data, no_of_days):
    finaldf_list = []

    for speed_reduction in range(0, 31):
        finaldf = find_cii_prediction(subset_port_data, subset_sail_data, no_of_days, speed_reduction)
        finaldf['Speed Reduction (%)'] = speed_reduction

        for i, row in finaldf.iterrows():
            reduced_sail_co2 = row['Sail_CO2_value']
            reduced_port_co2 = row['Port_CO2_Value'] + row['Port_CO2']
            reduced_co2 = reduced_sail_co2 + reduced_port_co2  

            attained_cii = (reduced_co2 * 1000000) / (row['deadweight'] * (row['distance'] + row['Sail_Distance_Value']))
            A_R_Ratio = round((attained_cii / row['required_cii_2024']), 2)
            cii_rating_after_reduction = cii_rating(A_R_Ratio)

            finaldf.at[i, 'attained_cii'] = attained_cii
            finaldf.loc[i, 'A_R_Ratio'] = A_R_Ratio
            finaldf.loc[i, 'cii_rating'] = cii_rating_after_reduction

        finaldf_list.append(finaldf)

    all_results_df = pd.concat(finaldf_list, keys=range(0, 31))
    
    return all_results_df

# Plotting the results
final_results_df = find_cii_for_speed_reductions(subset_port_df, subset_sail_df, 65)

fig = go.Figure()

fig.add_trace(go.Bar(
    x=final_results_df['Speed Reduction (%)'],
    y=final_results_df['A_R_Ratio'],
    name='Avg CII Ratio',
    marker_color='orange'
))

fig.add_trace(go.Scatter(
    x=final_results_df['Speed Reduction (%)'],
    y=final_results_df['A_R_Ratio'],
    mode='lines+markers+text',
    name='Avg CII Ratio Line',
    text=final_results_df['A_R_Ratio'].round(2),
    textposition='top center',
    line=dict(color='blue')
))

fig.add_hline(y=1.19, line_dash="dash", line_color="red")
fig.add_hline(y=1.07, line_dash="dash", line_color="orange")
fig.add_hline(y=1, line_dash="dash", line_color="yellow")
fig.add_hline(y=0.94, line_dash="dash", line_color="lightgreen")
fig.add_hline(y=0.83, line_dash="dash", line_color="green")

fig.update_yaxes(range=[0.6, 1.4])
fig.update_layout(
    xaxis_title="Speed Reduction (%)",
    yaxis_title="CII Ratio (Attained/Required)",
    legend=dict(
        orientation="h",
        yanchor="bottom",
        y=1.02,
        xanchor="center",
        x=0.5
    ),
    height=600,
    width=1000
)

fig.show()
```

```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go

def predict_co2_port(subset_port_data, port_co2_model):
    predicted_co2_port = {}
    for cluster in subset_port_data['Cluster'].unique():
        df = subset_port_data[subset_port_data['Cluster'] == cluster][['mean_draft', 'ae_t_steaming', 'aux_running', 'blr_running', 'activity_time']]
        port_co2_predicted = port_co2_model.predict(df).sum()
        predicted_co2_port[f'Port_CO2_{cluster}'] = round(port_co2_predicted, 2)
    return predicted_co2_port

def predict_co2_sail(subset_sail_data, sail_co2_model, speed_reduction):
    predicted_co2_sail = {}
    for cluster in subset_sail_data['Cluster'].unique():
        df = subset_sail_data[subset_sail_data['Cluster'] == cluster][['speed', 'mean_draft', 'new_ss', 'me_actual_steaming_time', 'ae_t_steaming', 'aux_running', 'blr_running']]
        df['speed'] = df['speed'] - df['speed'] * speed_reduction / 100
        sail_co2_predicted = sail_co2_model.predict(df).sum()
        predicted_co2_sail[f'Sail_CO2_{cluster}'] = round(sail_co2_predicted, 2)
    return predicted_co2_sail

def calculate_distance_sail(subset_sail_data, sail_distance_model, speed_reduction):
    predicted_distance_sail = {}
    for cluster in subset_sail_data['Cluster'].unique():
        df = subset_sail_data[subset_sail_data['Cluster'] == cluster][['speed', 'mean_draft', 'new_ss', 'me_actual_steaming_time', 'ae_t_steaming', 'aux_running', 'blr_running']]
        df['speed'] = df['speed'] - df['speed'] * speed_reduction / 100
        sail_distance_predicted = sail_distance_model.predict(df).sum()
        predicted_distance_sail[f'Sail_Distance_{cluster}'] = round(sail_distance_predicted, 2)
    return predicted_distance_sail

def find_cii_prediction(subset_port_data, subset_sail_data, no_of_days, speed_reduction, port_co2_model, sail_co2_model, sail_distance_model):
    predicted_co2_port = predict_co2_port(subset_port_data, port_co2_model)
    predicted_co2_sail = predict_co2_sail(subset_sail_data, sail_co2_model, speed_reduction)
    predicted_distance_sail = calculate_distance_sail(subset_sail_data, sail_distance_model, speed_reduction)

    finaldata = []
    for sail_key, sail_val in predicted_co2_sail.items():
        for port_key, port_val in predicted_co2_port.items():
            dist_key = sail_key.replace('CO2', 'Distance')
            dist_val = predicted_distance_sail.get(dist_key, 0)
            finaldata.append((sail_key, sail_val, port_key, port_val, dist_key, dist_val))

    finaldf = pd.DataFrame(finaldata, columns=['Sail_CO2', 'Sail_CO2_value', 'Port_CO2', 'Port_CO2_Value', 'Sail_Distance', 'Sail_Distance_Value'])
    finaldf[['Sail_CO2_value', 'Port_CO2_Value', 'Sail_Distance_Value']] = finaldf[['Sail_CO2_value', 'Port_CO2_Value', 'Sail_Distance_Value']].astype(float).round(2)

    finaldf['deadweight'] = 39228
    finaldf['required_cii_2024'] = 10.46
    finaldf['attained_cii'] = (finaldf['Sail_CO2_value'] + finaldf['Port_CO2_Value']) * 1_000_000 / (finaldf['deadweight'] * finaldf['Sail_Distance_Value'])
    finaldf['A_R_Ratio'] = (finaldf['attained_cii'] / finaldf['required_cii_2024']).round(2)
    finaldf['cii_rating'] = finaldf['A_R_Ratio'].apply(cii_rating)
    finaldf['avg_speed'] = (finaldf['Sail_Distance_Value'] / (no_of_days * 24)).round(2)

    return finaldf
```
