```python
data=pd.read_csv("NOONDATA_WEC_VAN_RIJIN.csv")
```

```python
df.dtypes #checking the variable types
```

```python

df['REPORT_DATE_TIME'] = pd.to_datetime(df['REPORT_DATE_TIME'], format='%d/%m/%Y %H:%M', errors='coerce')

# Extract the date and time components
df['Date'] = df['REPORT_DATE_TIME'].dt.date
df['Time'] = df['REPORT_DATE_TIME'].dt.time

# Convert the 'Date' column to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Filter for the year 2021
cii_year = 2023
Res_df = df[df['Date'].dt.year == cii_year]
Res_df.set_index('Date', inplace=True)
```

```python
Res_df.head()
```

```python
Res_df = Res_df.sort_values(by='REPORT_DATE_TIME')
```

```python

start_date = '2023-06-18'
end_date = '2023-07-18'
start_date = pd.to_datetime(start_date, format='%Y-%m-%d')
end_date = pd.to_datetime(end_date, format='%Y-%m-%d')

filtered_df = Res_df[(Res_df['REPORT_DATE_TIME'] >= start_date) & (Res_df['REPORT_DATE_TIME'] <= end_date)]


```

```python
filtered_df.head()
```

```python
pd.set_option('display.max_columns', None)  
filtered_df.tail()
```

```python
filtered_df['Total_Calculated_MGO_LS'] = filtered_df[['FUEL_M_E_MGO_LS', 'FUEL_AUX_MGO_LS', 'FUEL_BOILER_MGO_LS']].sum(axis=1)

filtered_df['Total_Calculated_ULS'] = filtered_df[['FUEL_M_E_Dstlt_ULS', 'FUEL_Aux_Dstlt_ULS', 'FUEL_Boiler_Dstlt_ULS']].sum(axis=1)

filtered_df['Total_fuel'] = filtered_df[['Total_Calculated_MGO_LS', 'Total_Calculated_ULS']].sum(axis=1)

filtered_df['Total_BIO_Calculated'] = (filtered_df['Total_Calculated_MGO_LS'] * 2.417) + (filtered_df['Total_Calculated_ULS'] * 2.567)
```

```python
dist = filtered_df['MILES_BY_GPS'].sum()
dist
```

```python
dwt= 13734
filtered_df = filtered_df.assign(DWT=dwt)
```

```python
TAFC= filtered_df['Total_BIO_Calculated'].sum() * 10**6
print(TAFC)
```

# ATTAINED CII

```python
ATT_CII = TAFC / (dist * dwt)
ATT_CII
```
