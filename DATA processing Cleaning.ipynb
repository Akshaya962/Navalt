```python
import string
# Convert 'IMO No' column to strings
cf['IMO No'] = cf['IMO No'].astype(str)

# Translation table
translator = str.maketrans('', '', string.punctuation)

# Remove punctuation using translation table
cf['IMO No'] = cf['IMO No'].apply(lambda x: x.translate(translator))

# Convert cleaned strings back to integers
cf['IMO No'] = cf['IMO No'].astype(int)
```

```python
import pandas as pd

# Load vessel details (charter_data) from Excel
charter_data = pd.read_excel(r'C:\Users\akshaya\Downloads\charter.xlsx')

# Assuming df_sorted is your charter data, already loaded
# Clean vessel names by stripping whitespace, converting to uppercase, and removing non-alphanumeric characters for comparison
charter_data['Name'] = charter_data['Name'].str.strip().str.upper().str.replace(r'[^A-Z0-9 ]', '', regex=True)
df_sorted['VSL'] = df_sorted['VSL'].str.strip().str.upper().str.replace(r'[^A-Z0-9 ]', '', regex=True)

# Clean IMO columns by converting them to strings (in case they are numbers)
charter_data['IMO'] = charter_data['IMO'].astype(str).str.strip()
df_sorted['IMO No'] = df_sorted['IMO No'].astype(str).str.strip()

# Step 1: Check if the IMO numbers match at all
print("Number of unique IMO No in charter data:", df_sorted['IMO No'].nunique())
print("Number of unique IMO in vessel data:", charter_data['IMO'].nunique())

# Step 2: Ensure there is overlap between the two IMO columns
common_imos = set(df_sorted['IMO No']).intersection(set(charter_data['IMO']))
print(f"Number of matching IMO numbers: {len(common_imos)}")

if len(common_imos) == 0:
    print("No matching IMO numbers found between the two datasets.")
else:
    print("Matching IMO numbers exist, proceeding with further checks.")

# Step 3: Merge the two datasets on the IMO columns
merged_data = pd.merge(df_sorted, charter_data, how='left', left_on='IMO No', right_on='IMO')

# Step 4: Identify rows where names don't match
merged_data['Name_Match'] = merged_data['VSL'] == merged_data['Name']

# Filter for incorrect matches (where IMO matches but names don't)
incorrect_data = merged_data[(merged_data['Name_Match'] == False) & merged_data['IMO'].notna()]

# Save the incorrect matches if needed
incorrect_data.to_csv('incorrect_data.csv', index=False)

# Print incorrect matches
print("Incorrect name matches:")
print(incorrect_data[['IMO No', 'VSL', 'Name']])

# Step 5: Diagnose unmatched data
# Print IMOs in charter data that don't have a match in vessel data
unmatched_imos = df_sorted[~df_sorted['IMO No'].isin(charter_data['IMO'])]
print("IMOs in charter data not found in vessel data:")
print(unmatched_imos[['IMO No', 'VSL']].head(10))

# Print IMOs in vessel data that don't have a match in charter data
unmatched_vessels = charter_data[~charter_data['IMO'].isin(df_sorted['IMO No'])]
print("IMOs in vessel data not found in charter data:")
print(unmatched_vessels[['IMO', 'Name']].head(10))
```
