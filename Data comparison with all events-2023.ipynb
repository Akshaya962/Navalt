```python
imo_df['Date_UTC'] = pd.to_datetime(imo_df['Date_UTC'])

result = imo_df.groupby('IMO')['Date_UTC'].agg(['min', 'max']).reset_index()

# Calculate number of days
result['min'] = pd.to_datetime(result['min'])
result['max'] = pd.to_datetime(result['max'])
result['Days Count'] = (result['max'] - result['min']).dt.days + 1

missing_dates = []

for _, row in result.iterrows():
    imo_no = row['IMO']
    date_range = pd.date_range(start=row['min'], end=row['max'], freq='D')
    actual_dates = set(imo_df[imo_df['IMO'] == imo_no]['Date_UTC'])
    missing = list(set(date_range) - actual_dates)
    missing_str = [date.strftime('%Y-%m-%d') for date in missing]  # Convert datetime.date objects to string
    missing_dates.append([imo_no, missing_str, len(missing)])


missing_dates_imodcs = pd.DataFrame(missing_dates, columns=['imo', 'Missing Dates', 'Count'])

print(missing_dates_imodcs)
```

```python
missing_dates_imodcs.to_csv('MIMO-2023.csv')
```

```python
import pandas as pd

# Assuming 'fuel_imodcs' is your DataFrame containing fuel data
fuel_imodcs['Date_UTC'] = pd.to_datetime(fuel_imodcs['Date_UTC'])

# Dropping duplicate rows based on 'IMO' and 'Date_UTC' columns
fuel_imodcs.drop_duplicates(subset=['IMO', 'Date_UTC'], inplace=True)

# Assigning a unique number to each date within each vessel's data
fuel_imodcs['Days_running_imodcs'] = fuel_imodcs.groupby('IMO').cumcount() + 1

# Modifying the assigned day numbers to ensure they don't exceed 365
fuel_imodcs['Days_running_imodcs'] = fuel_imodcs['Days_running_imodcs'].apply(lambda x: x % 365 if x <= 365 else x % 365 + 1)

# Grouping by 'IMO' and summing the modified assigned numbers to calculate days run
grouped_dates = fuel_imodcs.groupby('IMO')['Days_running_imodcs'].agg(Days_running_imodcs='max')

# Calculating the count of missing days for each vessel
grouped_dates['Missing_days_imodcs'] = 365 - grouped_dates['Days_running_imodcs']

# Extracting count of unique dates for each IMO
count_unique_dates_per_imo = fuel_imodcs.groupby('IMO')['Date_UTC'].nunique().reset_index()
count_unique_dates_per_imo.columns = ['IMO', 'Count_Unique_Dates']

# Merging the two DataFrames on 'IMO'
merged_data = pd.merge(grouped_dates, count_unique_dates_per_imo, on='IMO')

merged_data.head(7)
```
