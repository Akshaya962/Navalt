```python
import numpy as np
import pandas as pd
import warnings
import matplotlib.pyplot as plt
import requests
import io

warnings.filterwarnings("ignore")
```

```python
MSC_noon_2022=pd.read_csv("NOONDATA (2) 1.csv")
```

```python
pd.set_option('display.max_columns', None)  
MSC_noon_2022
```

```python
MSC_noon_2022['Date'] = pd.to_datetime(MSC_noon_2022['Corrected_Date']).apply(lambda x:x.strftime("%d"))
MSC_noon_2022['Month'] = pd.to_datetime(MSC_noon_2022['Corrected_Date']).apply(lambda x:x.strftime("%m"))
MSC_noon_2022['Year'] = pd.to_datetime(MSC_noon_2022['Corrected_Date']).apply(lambda x:x.strftime("%Y"))
MSC_noon_2022['Month_name'] = pd.to_datetime(MSC_noon_2022['Corrected_Date']).apply(lambda x:x.strftime("%B"))
MSC_noon_2022['date'] = pd.to_datetime(MSC_noon_2022["Corrected_Date"]).dt.date
MSC_noon_2022.dropna(subset=['IMO_No'])
MSC_noon_2022_sort=MSC_noon_2022.sort_values(by="Corrected_Date")
#Total HS,LS,MGO and Total Fuel
MSC_noon_2022_sort['Total_Calculated_HS']= MSC_noon_2022_sort[['FUEL_M_E_Rsdl_HS' , 'FUEL_Aux_Rsdl_HS' , 'FUEL_Boiler_Rsdl_HS']].sum(axis=1)
MSC_noon_2022_sort['Total_Calculated_LS'] = MSC_noon_2022_sort[['FUEL_M_E_Rsdl_VLS' , 'FUEL_M_E_Dstlt_VLS' , 'FUEL_M_E_Tnktnr_Dstlt_VLS' ,'FUEL_Aux_Rsdl_VLS' ,  'FUEL_Aux_Dstlt_VLS' , 'FUEL_Aux_Tnktnr_Dstlt_VLS','FUEL_Boiler_Rsdl_VLS','FUEL_Boiler_Dstlt_VLS','FUEL_Boiler_Tnktnr_Dstlt_VLS']].fillna(0).sum(axis=1)
MSC_noon_2022_sort['Total_Calculated_ULS'] = MSC_noon_2022_sort[['FUEL_M_E_Rsdl_ULS' , 'FUEL_Aux_Rsdl_ULS' , 'FUEL_Boiler_Rsdl_ULS' ,'FUEL_M_E_Dstlt_ULS' ,'FUEL_Aux_Dstlt_ULS' , 'FUEL_Boiler_Dstlt_ULS']].fillna(0).sum(axis=1)
MSC_noon_2022_sort['Total_Calculated_Fuel'] = MSC_noon_2022_sort[['Total_Calculated_HS' ,'Total_Calculated_LS' , 'Total_Calculated_ULS']].fillna(0).sum(axis=1)

#HS_CO2,LSCO2,MGO_CO2 and Total CO2
MSC_noon_2022_sort['HS_CO2'] = 3.114 * MSC_noon_2022_sort["Total_Calculated_HS"]
MSC_noon_2022_sort['LS_CO2'] = 3.151 * MSC_noon_2022_sort["Total_Calculated_LS"]
MSC_noon_2022_sort['ULS_CO2'] = 3.206 * MSC_noon_2022_sort["Total_Calculated_ULS"]
MSC_noon_2022_sort['Total_CO2'] = MSC_noon_2022_sort[['HS_CO2' ,'LS_CO2' ,'ULS_CO2']].fillna(0).sum(axis=1)

MSC_noon_2022_sort["fuel_m_e_calculated"]=MSC_noon_2022_sort[["FUEL_M_E_Rsdl_HS","FUEL_M_E_Rsdl_VLS","FUEL_M_E_Dstlt_VLS",'FUEL_M_E_Rsdl_ULS','FUEL_M_E_Tnktnr_Dstlt_VLS']].fillna(0).sum(axis=1)
MSC_noon_2022_sort["fuel_aux_calculated"]=MSC_noon_2022_sort[["FUEL_Aux_Rsdl_VLS","FUEL_Aux_Rsdl_HS","FUEL_Aux_Dstlt_VLS",'FUEL_Aux_Rsdl_ULS','FUEL_Aux_Tnktnr_Dstlt_VLS']].fillna(0).sum(axis=1)
MSC_noon_2022_sort["fuel_boiler_calculated"]=MSC_noon_2022_sort[["FUEL_Boiler_Rsdl_HS","FUEL_Boiler_Dstlt_VLS","FUEL_Boiler_Rsdl_VLS",'FUEL_Boiler_Rsdl_ULS','FUEL_Boiler_Tnktnr_Dstlt_VLS']].fillna(0).sum(axis=1)
MSC_noon_2022_sort["Total_fuel_calculated"]=MSC_noon_2022_sort[["fuel_m_e_calculated","fuel_aux_calculated","fuel_boiler_calculated"]].fillna(0).sum(axis=1)
```

```python

MSC_noon_2022_str = MSC_noon_2022_sort[MSC_noon_2022_sort['VOY'].apply(lambda x: isinstance(x, str))]
MSC_noon_2022_str['VOY'] = MSC_noon_2022_str['VOY'].str.lstrip('0').str.rstrip('.0')

# Filter by Float Type and Convert to String
MSC_noon_2022_float = MSC_noon_2022_sort[MSC_noon_2022_sort['VOY'].apply(lambda x: isinstance(x, float))]
MSC_noon_2022_float['VOY'] = pd.to_numeric(MSC_noon_2022_float['VOY'], errors='coerce').fillna(0).astype(int).astype(str)

# Filter by Integer Type and Convert to String
MSC_noon_2022_int = MSC_noon_2022_sort[MSC_noon_2022_sort['VOY'].apply(lambda x: isinstance(x, int))]
MSC_noon_2022_int['VOY'] = MSC_noon_2022_int['VOY'].astype(int).astype(str)

# Concatenate DataFrames and Sort
MSC_noon_2022_sort_new = pd.concat([MSC_noon_2022_str, MSC_noon_2022_float, MSC_noon_2022_int])
MSC_noon_2022_sort_new = MSC_noon_2022_sort_new.sort_values(by=["Corrected_Date"])
```

# 3.  Changing unit of Fuel

```python

# Fuel calculations
fuel_hs_2022 = MSC_noon_2022_sort.groupby(["IMO_No"])[["Total_Calculated_HS"]].sum().reset_index()
Total_fuel_hs_2022 = fuel_hs_2022[["Total_Calculated_HS"]].sum().reset_index()
Total_fuel_hs_2022 = Total_fuel_hs_2022[0] / (10 ** 6)
Total_fuel_hs_2022.reset_index()

fuel_ls_2022 = MSC_noon_2022_sort.groupby(["IMO_No"])[["Total_Calculated_LS"]].sum().reset_index()
Total_fuel_ls_2022 = fuel_ls_2022[["Total_Calculated_LS"]].sum().reset_index()
Total_fuel_ls_2022 = Total_fuel_ls_2022[0] / (10 ** 6)
Total_fuel_ls_2022.reset_index()

fuel_uls_2022 = MSC_noon_2022_sort.groupby(["IMO_No"])[["Total_Calculated_ULS"]].sum().reset_index()
Total_fuel_uls_2022 = fuel_uls_2022[["Total_Calculated_ULS"]].sum().reset_index()
Total_fuel_uls_2022 = Total_fuel_uls_2022[0] / (10 ** 6)
Total_fuel_uls_2022.reset_index()


fuel_me_2022 = MSC_noon_2022_sort.groupby(["IMO_No"])[["fuel_m_e_calculated"]].sum().reset_index()
Total_fuel_me_2022 = fuel_me_2022[["fuel_m_e_calculated"]].sum().reset_index()
Total_fuel_me_2022 = Total_fuel_me_2022[0] / (10 ** 6)
Total_fuel_me_2022.reset_index()

fuel_aux_2022 = MSC_noon_2022_sort.groupby(["IMO_No"])[["fuel_aux_calculated"]].sum().reset_index()
Total_fuel_aux_2022 = fuel_aux_2022[["fuel_aux_calculated"]].sum().reset_index()
Total_fuel_aux_2022 = Total_fuel_aux_2022[0] / (10 ** 6)
Total_fuel_aux_2022.reset_index()

fuel_boiler_2022 = MSC_noon_2022_sort.groupby(["IMO_No"])[["fuel_boiler_calculated"]].sum().reset_index()
Total_fuel_boiler_2022 = fuel_boiler_2022[["fuel_boiler_calculated"]].sum().reset_index()
Total_fuel_boiler_2022 = Total_fuel_boiler_2022[0] / (10 ** 6)
Total_fuel_boiler_2022.reset_index()

fuel_2022 = MSC_noon_2022_sort.groupby(["IMO_No"])[["Total_fuel_calculated"]].sum().reset_index()
Total_fuel_2022 = fuel_2022[["Total_fuel_calculated"]].sum().reset_index()
Total_fuel_2022 = Total_fuel_2022[0] / (10 ** 6)
Total_fuel_2022.reset_index()

# CO2 calculations
fuel_CO2_2022 = MSC_noon_2022_sort.groupby(["IMO_No"])[["Total_CO2"]].sum().reset_index()
Total_CO2_2022 = fuel_CO2_2022[["Total_CO2"]].sum().reset_index()
Total_CO2_2022 =Total_CO2_2022[0] / (10 ** 6)
Total_CO2_2022.reset_index()
```

```python
#Steaming Time


Steam_2022=MSC_noon_2022_sort.groupby(["IMO_No"])[["TOTAL_STEAMING_TIME"]].sum().reset_index()
Total_Steam_2022=Steam_2022[["TOTAL_STEAMING_TIME"]].sum().reset_index()
Total_Steam_2022=Total_Steam_2022[0]/(24)
Total_Steam_2022.reset_index(drop=True, inplace=True)
```
