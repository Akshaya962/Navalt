```python
import pandas as pd

def get_corrected_date(report_date_time, offset_str):
    # Convert 'REPORT DATE TIME' to datetime
    report_date_time = pd.to_datetime(report_date_time, errors='coerce')
    
    # Extract hours and minutes from the offset
    sign = offset_str[3]  # '+' or '-'
    hours = int(offset_str[4:6])
    minutes = int(offset_str[7:9])
    
    # Adjust the offset: '+' means subtract, '-' means add
    offset = pd.Timedelta(hours=hours, minutes=minutes * (1 if sign == '+' else -1))
    
    # Calculate the corrected date
    corrected_datetime = report_date_time - offset  # Subtract for '+' and add for '-'
    
    # Return only the date part
    return corrected_datetime.date()



# Apply the function to calculate the corrected date
charter['Corrected_Date'] = charter.apply(lambda row: get_corrected_date(row['REPORT_DATE_TIME'], row['REPORT_DATE_TIME_Offset']), axis=1)


```

```python
charter['Corrected_Date'] = pd.to_datetime(charter['Corrected_Date'], format='%y/%m/%d %H:%M:%S').dt.strftime('%Y-%m-%d %H:%M:%S')
charter['Corrected_Date']
```

```python
charter['REPORT_DATE_TIME'] = charter['REPORT_DATE_TIME'].str.replace(r'\s+\+\d{2}:\d{2}$', '', regex=True)  # For positive timezone
charter['REPORT_DATE_TIME'] = charter['REPORT_DATE_TIME'].str.replace(r'\s+-\d{2}:\d{2}$', '', regex=True)  # For negative timezone

# Remove any leading or trailing whitespace
charter['REPORT_DATE_TIME'] = charter['REPORT_DATE_TIME'].str.strip()

# Convert back to datetime
charter['REPORT_DATE_TIME'] = pd.to_datetime(charter['REPORT_DATE_TIME'])

charter
```
