```python
pd.set_option('display.max_columns', None)
mdf=pd.read_csv('MTM_Imo.csv')
mdf.head(71)
```

```python
df = df.merge(mdf, left_on='Vessel_Name', right_on='id', how='inner')
```

```python
df.head()
```

```python
df = df.drop(['mail_date', 'IMO_No','VOY','Vessel_Name','Voyage_Code'], axis=1)

desired_columns = ['Vessel','id', 'IMO.No'] + [col for col in df.columns if col not in ['Vessel','id','IMO.No']]
df = df[desired_columns]
```

```python
df.tail()
```

```python
df['REPORT_DATE_TIME'] = pd.to_datetime(df['REPORT_DATE_TIME'], format='%d-%m-%y %H:%M')
df['Day'] = df['REPORT_DATE_TIME'].dt.day
df['Date'] = df['REPORT_DATE_TIME'].dt.date
df['Date'] = pd.to_datetime(df['Date'])
df['Year'] = df['REPORT_DATE_TIME'].dt.year
df['Month'] = df['REPORT_DATE_TIME'].dt.month
df['Hours'] = df['REPORT_DATE_TIME'].dt.hour
df['Minutes'] = df['REPORT_DATE_TIME'].dt.minute
df['Seconds'] = df['REPORT_DATE_TIME'].dt.second
df['DayOfWeek'] = df['Date'].dt.day_name()
df['Month_name'] = df['Date'].dt.strftime('%B')
df['Minutes_in_hours'] = df['Minutes'] / 60
df['Seconds_in_hours'] = df['Seconds'] / 3600
df['Total_Time'] = df['Hours'] + df['Minutes_in_hours'] + df['Seconds_in_hours']
df['Total_Time'] = df['Total_Time'].round(2)

df[['REPORT_DATE_TIME', 'Day', 'Date', 'Year', 'Month', 'Hours', 'Minutes', 'Seconds', 'DayOfWeek','Month_name','Total_Time']]
```

```python
df = df.sort_values(by=['Vessel', 'VoyageNo', 'REPORT_DATE_TIME']).reset_index(drop=True)
```

```python
df['New_Report_Type'] = df['Report_Type'].shift(1).fillna('') + '-' + df['Report_Type']
```
