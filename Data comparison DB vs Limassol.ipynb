```python
result['min'] = pd.to_datetime(result['min'])
result['max'] = pd.to_datetime(result['max'])
result['Days Count'] = (result['max'] - result['min']).dt.days+1
```

```python
db['Date_UTC'] = pd.to_datetime(db['Date_UTC'])
```

```python
missing_dates = []

for index, row in result.iterrows():
    imo_no = row['IMO']
    date_range = pd.date_range(start=row['min'], end=row['max'], freq='D')
    actual_dates = db[db['IMO'] == imo_no]['Date_UTC'].dt.floor('D').unique()
    missing_dates.append({
        'imo': imo_no,
        'Missing Dates': [date.strftime('%Y-%m-%d') for date in list(set(date_range) - set(actual_dates))],
        'Count': len(set(date_range) - set(actual_dates))  # Calculate count of unique missing dates
    })

missing_dates_df = pd.DataFrame(missing_dates)

print(missing_dates_df)
```

```python
lima['Total_HS_lima']=lima[['FOC_ME_RES_HS','FOC_AE_RES_HS','FOC_BLR_RES_HS']].sum(axis=1)
lima['Total_LS_lima']=lima[['FOC_ME_RES_ULS','FOC_ME_RES_VLS','FOC_ME_RES_ULS','FOC_ME_RES_VLS','FOC_BLR_RES_VLS','FOC_BLR_RES_ULS']].sum(axis = 1) 
lima['Total_UlS_lima']=lima[['FOC_ME_DIS','FOC_ME_DIS_ULS','FOC_ME_DIS_VLS','FOC_AE_DIS','FOC_AE_DIS_ULS','FOC_AE_DIS_VLS','FOC_BLR_DIS','FOC_BLR_DIS_ULS','FOC_BLR_DIS_VLS']].sum(axis = 1) 
```

```python
lima['ME_hs_lima'] =  lima[['FOC_ME_RES_HS']]
lima['AUX_hs_lima'] =  lima[['FOC_AE_RES_HS']]
lima['Boiler_hs_lima'] =  lima[['FOC_BLR_RES_HS']]
```

```python
lima['ME_ls_lima'] =  lima[['FOC_ME_RES_ULS','FOC_ME_RES_VLS']].sum(axis = 1) 
lima['AUX_ls_lima'] =  lima[['FOC_ME_RES_ULS','FOC_ME_RES_VLS']].sum(axis = 1)  
lima['Boiler_ls_lima'] =  lima[['FOC_BLR_RES_VLS','FOC_BLR_RES_ULS']] .sum(axis = 1) 
```

```python
lima['ME_uls_lima'] =  lima[['FOC_ME_DIS','FOC_ME_DIS_ULS','FOC_ME_DIS_VLS']].sum(axis = 1) 
lima['AUX_uls_lima'] =  lima[['FOC_AE_DIS','FOC_AE_DIS_ULS','FOC_AE_DIS_VLS']].sum(axis = 1) 
lima['Boiler_uls_lima'] =  lima[['FOC_BLR_DIS','FOC_BLR_DIS_ULS','FOC_BLR_DIS_VLS']] .sum(axis = 1) 
```

```python
lima['Total_ME_lima'] = lima[['ME_hs_lima','ME_ls_lima','ME_uls_lima']].sum(axis = 1)   
lima['Total_AUX_lima'] = lima[['AUX_hs_lima','AUX_ls_lima','AUX_uls_lima']].sum(axis = 1) 
lima['Total_Boiler_lima'] = lima[['Boiler_hs_lima','Boiler_ls_lima','Boiler_uls_lima']].sum(axis = 1) 
```
