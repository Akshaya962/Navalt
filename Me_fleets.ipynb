```python
import pandas as pd
```

```python
data = pd.read_excel('Vessel addition data 2023-24 1.xlsx')
```

```python
data
```

```python
df=data
```

```python
month_mapping = {
    'JAN': 'January',
    'FEB': 'February',
    'MARCH': 'March',
    'APRIL': 'April',
    'MAY': 'May',
    'JUN': 'June',
    'JULY': 'July',
    'AUG': 'August',
    'SEP': 'September',
    'OCT': 'October',
    'NOV': 'November',
    'DEC': 'December'
}

# Replace the month names using the mapping
df['Month'] = df['Month'].replace(month_mapping)

```

```python
df = df.fillna(0)

df
```

```python
import matplotlib.pyplot as plt
import plotly.express as px

# Matplotlib plot
plt.figure(figsize=(10, 6))
plt.plot(df['Month'], df['FLEET 1'], label='Fleet 1', marker='o')

plt.title('Fleet 1 Vessel Count Across Months')
plt.xlabel('Month')
plt.ylabel('Vessel Count')
plt.xticks(rotation=45)
plt.grid(True)

plt.legend()
plt.tight_layout()
plt.show()

# Plotly plot (interactive)
fig = px.line(df, x='Month', y='FLEET 1', title='Fleet 1 Vessel Count Across Months')
fig.update_xaxes(title_text='Month')
fig.update_yaxes(title_text='Vessel Count')
fig.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Assuming the month is a column and each fleet count is a column, drop Year if it's not necessary
df_clean = df.drop(columns=['Year'])  # Drop 'Year' if it's not needed for plotting

# Set the 'Month' as the index (y-axis) and fleets as columns (x-axis)
df_clean.set_index('Month', inplace=True)

# Plot the heatmap
plt.figure(figsize=(16, 10))
sns.heatmap(df_clean.T, cmap="YlGnBu", annot=True, fmt='.0f', cbar_kws={'label': 'Vessel Count'})
plt.title('Fleet Vessel Counts by Month', fontsize=16)
plt.xlabel('Fleet', fontsize=12)
plt.ylabel('Month', fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

fleet_columns = [col for col in df.columns if col.startswith('FLEET')]

fleet_totals = df[fleet_columns].sum()  
high_utilization_fleets = fleet_totals[fleet_totals > fleet_totals.mean()].index.tolist()

# 2. Fleets with Consistently High/Low Counts
fleet_means = df[fleet_columns].mean()  # Average vessels per fleet across months
fleet_std = df[fleet_columns].std()    # Standard deviation per fleet

high_fleets = fleet_means[fleet_means > fleet_means.mean()].index.tolist()
low_fleets = fleet_means[fleet_means < fleet_means.mean()].index.tolist()

# 3. Fleets with High Fluctuation
fluctuating_fleets = fleet_std[fleet_std > fleet_std.mean()].index.tolist()

print("### Analysis Results ###")
print("1. Fleets with High Utilization:")
print(high_utilization_fleets)

print("\n2. Fleets with Consistently High Vessel Counts:")
print(high_fleets)

print("\n3. Fleets with Consistently Low Vessel Counts:")
print(low_fleets)

print("\n4. Fleets with High Fluctuation:")
print(fluctuating_fleets)
```

```python
df = pd.DataFrame(data)

# Extract fleet columns
fleet_columns = [col for col in df.columns if col.startswith('FLEET')]

# Analysis
fleet_totals = df[fleet_columns].sum()
fleet_means = df[fleet_columns].mean()
fleet_std = df[fleet_columns].std()

# High Utilization Fleets
plt.figure(figsize=(12, 6))
sns.barplot(x=fleet_totals.index, y=fleet_totals.values, palette='viridis')
plt.title("High Utilization Fleets (Total Vessel Counts)")
plt.ylabel("Total Vessel Count")
plt.xlabel("Fleet")
plt.show()

# Consistently High/Low Fleets
plt.figure(figsize=(12, 6))
sns.barplot(x=fleet_means.index, y=fleet_means.values, palette='coolwarm')
plt.title("Consistently High/Low Fleets (Average Vessel Counts)")
plt.ylabel("Average Vessel Count")
plt.xlabel("Fleet")
plt.axhline(fleet_means.mean(), color='red', linestyle='--', label='Overall Average')
plt.legend()
plt.show()

# Fluctuating Fleets
plt.figure(figsize=(12, 6))
sns.barplot(x=fleet_std.index, y=fleet_std.values, palette='magma')
plt.title("Fluctuating Fleets (Standard Deviation of Vessel Counts)")
plt.ylabel("Standard Deviation")
plt.xlabel("Fleet")
plt.axhline(fleet_std.mean(), color='blue', linestyle='--', label='Overall Mean Fluctuation')
plt.legend()
plt.show()
```

```python
# Separate data for each year
df_2023 = df[df['Year'] == 2023]
df_2024 = df[df['Year'] == 2024]

# Plotting function
def plot_fleet_activity(df, year):
    df = df.set_index('Month').drop(columns=['Year', 'TOTAL VESSEL'])
    df.T.plot(kind='bar', figsize=(12, 8), stacked=True, colormap='viridis')
    plt.title(f"Fleet Activity in {year}")
    plt.xlabel("Fleet")
    plt.ylabel("Number of Vessels")
    plt.xticks(rotation=45)
    plt.legend(title="Month")
    plt.tight_layout()
    plt.show()

# Plot for 2023
plot_fleet_activity(df_2023, 2023)

# Plot for 2024
plot_fleet_activity(df_2024, 2024)
```

```python
def plot_consistently_high_low(df):
    avg_vessels = df.drop(columns=['Year', 'Month', 'TOTAL VESSEL']).mean()
    overall_avg = avg_vessels.mean()

    avg_vessels.plot(kind='bar', figsize=(12, 8), color='lightgreen', edgecolor='black')
    plt.axhline(overall_avg, color='red', linestyle='--', label=f'Overall Average ({overall_avg:.1f})')
    plt.title("Consistently High/Low Fleets (Average Vessel Counts)")
    plt.xlabel("Fleet")
    plt.ylabel("Average Number of Vessels")
    plt.xticks(rotation=45)
    plt.legend()
    plt.tight_layout()
    plt.show()

plot_consistently_high_low(df)
```
