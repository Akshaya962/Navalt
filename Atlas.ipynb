```python
import plotly.express as px
missing_dates_dict = {}

data['Timestamp (UTC)'] = pd.to_datetime(data['Timestamp (UTC)'], dayfirst=True)

reported_dates = data['Timestamp (UTC)'].dt.date.unique()

if len(reported_dates) > 0:
    start_date = reported_dates.min()
    end_date = reported_dates.max()
    
    date_range = pd.date_range(start=start_date, end=end_date).date
    
    missing = [date for date in date_range if date not in reported_dates]
    
    if missing:
        for date in missing:
            month_year = date.strftime('%b-%Y')  
            if month_year not in missing_dates_dict:
                missing_dates_dict[month_year] = [] 
            missing_dates_dict[month_year].append(date)

total_missing_dates = 0  

for month_year, missing_dates in missing_dates_dict.items():
    print(f"Month: {month_year}")
    print(f"Missing Count: {len(missing_dates)}")
    print(f"Missing Dates: {', '.join(map(str, missing_dates))}")
    print()
    
    total_missing_dates += len(missing_dates)

print(f"Total Missing Dates: {total_missing_dates}")

month_years = list(missing_dates_dict.keys())
missing_counts = [len(missing_dates_dict[month]) for month in month_years]

df = pd.DataFrame({
    'Month-Year': month_years,
    'Missing Dates Count': missing_counts
})

fig = px.bar(df, x='Month-Year', y='Missing Dates Count',
             title='Missing Dates Count per Month',
             labels={'Month-Year': 'Month-Year', 'Missing Dates Count': 'Missing Dates Count'},
             color='Missing Dates Count', color_continuous_scale='earth')

fig.update_layout(
    xaxis_title='Month-Year',
    yaxis_title='Missing Dates Count',
    xaxis_tickangle=45,
    coloraxis_showscale=False,
    template='plotly_white',
    xaxis=dict(showgrid=False),  
    yaxis=dict(showgrid=False)  
)

fig.show()
```

```python
def find_outliers(data, column, lower_bound, upper_bound):
    outliers = data[(data[column] > upper_bound) | (data[column] < lower_bound)][['Timestamp (UTC)', column]].copy()
    outliers['outlier'] = column
    outliers.rename(columns={column: 'value'}, inplace=True)
    return outliers

# List of columns with their bounds and labels
outlier_conditions = [
    ('Cargo Weight (t)', 0, 300000),
    ('Cargo TEU full', 0, 25000),
    ('Draft Forward (m)', 0, 30),
    ('Draft Aft (m)', 0, 30),
    ('Speed Over Ground (kn)', 0, 50),
    ('Speed Through Water (kn)', 0, 50),
    ('Total Power ME (AVG) (kW)', 0, 100000),
    ('AVG Shaft RPM (1/min)', 0, 750),
    ('Distance Over Ground (nm)', 0, 1000),
    ('AVG Slip Over Ground (%)', -100, 100),
    ('Sea State (Bft)', 0, 10),
    ('Douglas Sea Scale (Dgl)', 0, 10),
    ('Total Running Hours ME (hours)', 0, 26)
]

outliers_list = [find_outliers(data, col, lb, ub) for col, lb, ub in outlier_conditions]

all_outliers = pd.concat(outliers_list, ignore_index=True)

outliers = all_outliers[['Timestamp (UTC)', 'outlier', 'value']]

outliers
```

```python
outliers.to_excel('outliers.xlsx')
```
