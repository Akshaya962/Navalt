```python
Res_df.head()
```

# ATTAINED CII

```python
'''

ATTAINED CII= MACSS OF CO2(M)/ TRANSPORT WORK(W)
CII= [ANNUAL FUEL CONSUMPTION]*[CO2 EMISSION FACTOR]/ TRANSPORT WORK
REQUIRED CII= CII(REF)*[(100-Z)/100]

'''
```

```python
#Total HS
Res_df['Total_Calculated_HS'] =  Res_df[['FUEL_M_E_Rsdl_HS' , 'FUEL_Aux_Rsdl_HS' , 'FUEL_Boiler_Rsdl_HS']].sum(axis=1)

#Total LS
Res_df['Total_Calculated_LS'] =  Res_df[['FUEL_M_E_Rsdl_VLS' , 'FUEL_Aux_Rsdl_VLS' , 'FUEL_Boiler_Rsdl_VLS' ,'FUEL_M_E_Rsdl_ULS' ,  'FUEL_Aux_Rsdl_ULS' , 'FUEL_Boiler_Rsdl_ULS']].sum(axis=1)

#Total ULS
Res_df['Total_Calculated_ULS'] =  Res_df[['FUEL_M_E_Dstlt_VLS' , 'FUEL_Aux_Dstlt_VLS' , 'FUEL_Boiler_Dstlt_VLS' ,'FUEL_M_E_Dstlt_ULS' ,'FUEL_Aux_Dstlt_ULS' , 'FUEL_Boiler_Dstlt_ULS' ,'FUEL_M_E_Tnktnr_Dstlt_VLS' , 'FUEL_Aux_Tnktnr_Dstlt_VLS' ,'FUEL_Boiler_Tnktnr_Dstlt_VLS']].sum(axis=1)
```

```python
"""
EMISSION FACTOR:

FUEL OIL TYPE                CF
DIESEL/ GAS OIL             3.206
LIGHT FUEL OIL              3.151
HEAVY FUEL OIL              3.114
LPG (PROPANE)               3.000
LPG (BUTANE)                3.030
LNG                         2.750
METHANOL                    1.375
ETHANOL                     1.913
OTHER

"""
```

```python
#carbon emission factor
Res_df['CO2_HFC']=Res_df['Total_Calculated_HS']*3.114
Res_df['CO2_LFC']=Res_df['Total_Calculated_LS']*3.151
Res_df['CO2_UFC']=Res_df['Total_Calculated_ULS']*3.206
```

```python
Res_df['Mass_FC']=Res_df[['CO2_HFC','CO2_LFC','CO2_UFC']].sum(axis=1)
```

```python
# Transport of work :Distance
dist = Res_df['MILES_BY_GPS'].sum()
print(dist)
```

```python
#Deadweight
dwt= 127951.80
Res_df = Res_df.assign(DWT=dwt)
#Res_df.head()
```

```python
#Total annual fuel consumption
TAFC= Res_df['Mass_FC'].sum() * 10**6
print(TAFC)
```

```python
#Attained CII
ATT_CII = TAFC / (dist * dwt)
ATT_CII
```

# REFERENCE CII

```python
'''

CII(REF)= A.CAPACITY^-C

PARAMETER FOR A AND C ARE

SHIP TYPE                           A            C         CAPACITY
TANKER                             5247         0.610       DWT
CONTAIER SHIP                      1984         0.489       DWT
BULK CARRIER: DWT >= 2,79,000      4745         0.622       2,79,00     
BULK CARRIER: DWT <= 2,79,00       4745         0.622       DWT
COMBINATION CARRIER                40853        0.812       DWT

'''
```

```python
#Parameters for reference cii
a = int(input("A:"))
c = float(input("C:"))
dwt= 110652.40
```

```python
CII_REF = a * (dwt ** (-c))
print(CII_REF)
```

# REQUIRED CII

```python
'''
Required CII= CII(REF) * [(100-Z)/100]

REDUCATION FACTOR(Z%)
YEAR           REDUCTION FACTOR RELATIVE TO 2019
2019                   
2020                  1%
2021                  2%
2022                  3%
2023                  5%
2024                  7%
2025                  9%
2026                  11%

'''

```

```python
#Z-Reduction value
years = {
    2019: 0,
    2020: 1,
    2021: 2,
    2022: 3 ,
    2023: 5 ,
    2024: 7 ,
    2025: 9 ,
    2026: 11
}

a = int(input("Enter a year: "))    #ENTER THE CII YEAR
if a in years:
    Z = years[a]
    print(Z)
else:
    print("No Z value")
```

```python
#Required CII 
CII_REQ = CII_REF * ((100 - Z) / 100)
print(CII_REQ)
```

# CII RATING

```python
'''

CII RATIO= ATTAINED CII/ REQUIRED CII
                                                DD VECTORS
SHIP TYPE                          CAPACITY     D1     D2    D3    D4  
BULK CARRIER                         DWT       0.86   0.94  1.06  1.18
TANKER                               DWT       0.82   0.93  1.08  1.28
CONTAINER                            DWT       0.83   0.94  1.07  1.19
COMBINATION                          DWT       0.87   0.96  1.06  1.14
GENERAL CARGO SHIP                   DWT       0.83   0.94  1.06  1.19


CII RATIO <= D1 ---------- A RATING
D1<CII RATIO <= D2 ------- B RATING
D2<CII RATIO <= D3 ------- C RATING
D3<CII RATIO <= D4 ------- D RATING
CII RATIO  > D4 ---------- E RATING

A – Major superior performance level
B – Minor superior performance level
C – Moderate performance level
D – Minor Inferior performance level
E – Inferior performance level

All vessels must aspire to be at least C-rated.Any vessel that is rated D or E for three years in a row will have to 
submit a corrective action plan that explains the efforts that will be taken to achieve an acceptable rating (C or above). 
 
'''
```

```python
#CII Ratio
CII_RATIO= ATT_CII/ CII_REQ
print(CII_RATIO)
```

# RESULT 

```python
ships = {
    'BULK': {'D1': 0.86, 'D2': 0.94, 'D3': 1.06, 'D4': 1.18},
    'TANKER': {'D1': 0.82, 'D2': 0.93, 'D3': 1.08, 'D4': 1.28},
    'CONTAINER': {'D1': 0.83, 'D2': 0.94, 'D3': 1.07, 'D4': 1.19},
    'COMBINATION': {'D1': 0.87, 'D2': 0.96, 'D3': 1.06, 'D4': 1.14},
    'GENERAL CARGO SHIP': {'D1': 0.83, 'D2': 0.94, 'D3': 1.06, 'D4': 1.19},
}

def ratings(ship, cii_ratio):
    a = ships.get(ship)
    if a:
        if cii_ratio <= a['D1']:
            return 'A RATING'
        elif a['D1'] < cii_ratio <= a['D2']:
            return 'B RATING'
        elif a['D2'] < cii_ratio <= a['D3']:
            return 'C RATING'
        elif a['D3'] < cii_ratio <= a['D4']:
            return 'D RATING'
        else:
            return 'E RATING'
    else:
        return 'Data not founded'

ship = input("Enter the ship: ").strip().upper()  
CII_RATIO = CII_RATIO
rating = ratings(ship, CII_RATIO)
print(f'The rating for MSC AGREGENTO is: {rating}')
```
