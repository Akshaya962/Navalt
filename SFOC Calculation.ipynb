# SFOC Calculation

```python
#SFOC (g/kwh) = Mass of fuel consumed per hour / Power developed in KW
```

```python
fuel_columns = ['FUEL_M_E_HS', 'FUEL_M_E_LS', 'FUEL_M_E_MDO', 'FUEL_M_E_MGO_HS', 'FUEL_M_E_MGO_LS', 'FUEL_ME_VLSFOH', 'FUEL_ME_VLSFOL']
df[fuel_columns] = df[fuel_columns] * 1000000 

df['SFOC_Calc'] = (df[fuel_columns].sum(axis=1) / (df['TOTAL_STEAMING_TIME'] * df['POWER_kW'])).round(2)
df.head(10)
```

```python
import pandas as pd

data = {
    'TFC': [20,20,20,20],  
    'TOTAL_STEAMING_TIME': [24, 24, 14, 10],  
    'POWER_kW': [900, 850, 650, 690],  
}

sf = pd.DataFrame(data)

sf['TFC'] = sf['TFC'] * 1000000  

sf['SFOC_Calc'] = (sf['TFC']/ (sf['TOTAL_STEAMING_TIME'] * sf['POWER_kW'])).round(2)

print(sf)
```

```python

data = {
    'AE': ['AE 1', 'AE 2', 'AE 3', 'AE 4'],
    'TFC': [20,20,20,20],  
    'TOTAL_STEAMING_TIME': [24, 24, 14, 10], 
    'POWER_kW': [900, 850, 650, 690]
}

sf = pd.DataFrame(data)


sf['TFC'] = sf['TFC'] * 1000000  

sf['Fuel_Consumption'] = (sf['TFC'] / sf['TOTAL_STEAMING_TIME']).round(2)  

sf['SFOC'] = (sf['Fuel_Consumption'] / (sf['POWER_kW'] * 24)).round(2)

print(sf)
```
