```python
import pandas as pd
import numpy as np
data = pd.read_csv( "cii_daily_cristina_.csv")
data
```

```python
data['date'] = pd.to_datetime(data['date'], errors='coerce')
```

```python
data
```

```python
key_date_starting = pd.to_datetime('2023-10-11')
key_date_ending = pd.to_datetime('2023-12-18')
```

```python
key_date_starting
```

```python

data_1 = data[['imo',  'date',  'total_hs',
       'total_ls', 'total_uls', 'total_fuel', 'total_hs_co_2',
       'total_ls_co_2', 'total_uls_co_2', 'total_co_2', 
       'miles_by_gps', 'steaming_time', 'total_aux_engine_time',
       'speed_by_log', 'speed_by_gps','rpm', 'slip',
     'draft_aft', 'draft_fwd',  'dwt', 'attained_cii', 'current_cii'
       ,'cii_rating', 'required_cii']]
data_1['speed_cal'] = data_1['miles_by_gps']/data_1['steaming_time']
data_1['speed_cal'] = data_1['speed_cal'].replace([np.inf, -np.inf], np.nan)#When distance = 0 then speed become inf

data_1.isna().sum()
```

```python
before_6_month = key_date_starting - pd.DateOffset(months=4)
```

```python
df_starting = data_1[(data_1['date'] >= before_6_month) & (data_1['date'] < key_date_starting)]
df_starting.isna().sum()
```

```python
def mean_exclude_zeros(series):
    filtered = series[series != 0]
    return filtered.mean() if not filtered.empty else np.nan
```

```python
df_starting_gp = df_starting.groupby(['imo']).agg({'total_hs' :'sum','total_ls':'sum', 'total_uls':'sum', 'total_fuel':'sum', 'total_hs_co_2':'sum',
       'total_ls_co_2':'sum', 'total_uls_co_2':'sum', 'total_co_2':'sum', 
       'miles_by_gps':'sum', 'steaming_time':'sum', 'total_aux_engine_time':'sum',
       'speed_by_log':'mean', 'speed_by_gps':'mean','rpm':'mean', 'slip':'sum','dwt':'mean', 'required_cii':'mean','speed_cal':'mean',    'draft_aft': mean_exclude_zeros,'draft_fwd': mean_exclude_zeros}).reset_index().round(2)
df_starting_gp
```

```python
df_starting_gp['attained_cii'] = ((df_starting_gp['total_co_2']*1000000)/(df_starting_gp['miles_by_gps']*df_starting_gp['dwt'])).round(2)
#df_starting_gp['a_r'] = df_starting_gp['attained_cii']/df_starting_gp['required_cii']
df_starting_gp  = df_starting_gp[['imo', 'total_hs', 'total_ls', 'total_uls', 'total_fuel',
                              'total_co_2','miles_by_gps','required_cii','attained_cii','speed_cal',
                              'speed_by_log', 'speed_by_gps', 'rpm', 'slip', 'dwt', 'draft_aft','draft_fwd']]
df_starting_gp
```

```python
before_target_miles = df_starting_gp['miles_by_gps'][0]
before_target_miles
```

KEYDATES  ENDING 

```python
after_data = data_1[data_1['date'] > key_date_ending].sort_values(by='date')

cumulative_miles = 0
selected_rows = []

# Iterate through after_data to find rows summing up close to before_target_miles
for _, row in after_data.iterrows():
    cumulative_miles += row['miles_by_gps']
    selected_rows.append(row)
    if cumulative_miles >= before_target_miles:
        break

df_ending = pd.DataFrame(selected_rows)

# Determine the end date of the selected after data
end_date = df_ending['date'].max() if not df_ending.empty else pd.NaT

# Output the results
print(f"Target miles from before data: {before_target_miles}")
print(f"Cumulative miles from after data: {cumulative_miles}")
print(f"End date of the selected after data: {end_date}")
```

```python
df_ending_gp = df_ending.groupby(['imo']).agg({'total_hs' :'sum','total_ls':'sum', 'total_uls':'sum', 'total_fuel':'sum', 'total_hs_co_2':'sum',
       'total_ls_co_2':'sum', 'total_uls_co_2':'sum', 'total_co_2':'sum', 
       'miles_by_gps':'sum', 'steaming_time':'sum', 'total_aux_engine_time':'sum',
       'speed_by_log':'mean', 'speed_by_gps':'mean','rpm':'mean', 'slip':'sum','dwt':'mean', 'required_cii':'mean','speed_cal':'mean','draft_aft': mean_exclude_zeros,'draft_fwd': mean_exclude_zeros}).reset_index().round(2)
df_ending_gp
```

```python
df_ending_gp.columns.values
```

```python
df_ending_gp['attained_cii'] = ((df_ending_gp['total_co_2']*1000000)/(df_ending_gp['miles_by_gps']*df_ending_gp['dwt'])).round(2)
df_ending_gp  = df_ending_gp[['imo', 'total_hs', 'total_ls', 'total_uls', 'total_fuel',
                              'total_co_2','miles_by_gps','required_cii','attained_cii','speed_cal',
                              'speed_by_log', 'speed_by_gps', 'rpm', 'slip', 'dwt','draft_aft','draft_fwd']]
df_ending_gp
```

```python
import matplotlib.pyplot as plt

# Columns to be visualized in separate graphs
columns_group_1 = ['total_hs', 'total_ls', 'total_uls', 'total_fuel', 'total_co_2', 'miles_by_gps']
columns_group_2 = ['speed_by_log', 'speed_by_gps', 'draft_aft', 'draft_fwd', 'speed_cal', 'attained_cii']

# Filter the required columns for both groups
df_starting_group_1 = df_starting_gp[columns_group_1]
df_ending_group_1 = df_ending_gp[columns_group_1]

df_starting_group_2 = df_starting_gp[columns_group_2]
df_ending_group_2 = df_ending_gp[columns_group_2]

# Calculate mean values for each group
df_starting_mean_group_1 = df_starting_group_1.mean().values
df_ending_mean_group_1 = df_ending_group_1.mean().values

df_starting_mean_group_2 = df_starting_group_2.mean().values
df_ending_mean_group_2 = df_ending_group_2.mean().values

# Function to create a grouped bar chart
def plot_grouped_bar(x_labels, group1_values, group2_values, title, colors, group_labels):
    x = np.arange(len(x_labels))  # Label locations
    width = 0.35  # Width of the bars

    fig, ax = plt.subplots(figsize=(10, 6))

    # Bars for the first group
    bars1 = ax.bar(x - width / 2, group1_values, width, label=group_labels[0], color=colors[0])

    # Bars for the second group
    bars2 = ax.bar(x + width / 2, group2_values, width, label=group_labels[1], color=colors[1])

    # Add labels, title, and legend
    ax.set_xlabel('')
    ax.set_ylabel('Values')
    ax.set_title(title)
    ax.set_xticks(x)
    ax.set_xticklabels(x_labels, rotation=45, ha='right')
    ax.legend()

    # Add value annotations above bars with 90-degree rotation
    for bar in bars1:
        height = bar.get_height()
        ax.annotate(
            f'{height:.2f}', 
            xy=(bar.get_x() + bar.get_width() / 2, height),
            xytext=(0, 5),  # Position adjustment
            textcoords="offset points",
            ha='center', 
            va='bottom',
            rotation=45  # Rotate the label
        )

    for bar in bars2:
        height = bar.get_height()
        ax.annotate(
            f'{height:.2f}', 
            xy=(bar.get_x() + bar.get_width() / 2, height),
            xytext=(0, 5),  # Position adjustment
            textcoords="offset points",
            ha='center', 
            va='bottom',
            rotation=45  # Rotate the label
        )

    # Adjust layout and show the plot
    plt.tight_layout()
    plt.show()

# Plot for Group 1: ['total_fuel', 'total_co_2', 'miles_by_gps']
plot_grouped_bar(
    columns_group_1,
    df_starting_mean_group_1,
    df_ending_mean_group_1,
    title='Grouped Bar Chart: Pre and Post Analysis (Group 1)',
    colors=['blue', 'orange'],
    group_labels=['Pre Data', 'Post Data']
)

# Plot for Group 2: ['speed_by_log', 'speed_by_gps', 'attained_cii']
plot_grouped_bar(
    columns_group_2,
    df_starting_mean_group_2,
    df_ending_mean_group_2,
    title='Grouped Bar Chart: Pre and Post Analysis (Group 2)',
    colors=['green', 'red'],
    group_labels=['Pre Data', 'Post Data']
)
```
