```python
#use python remarks function next
def remark(data1):
    r1 = data1[data1['sulphur_content_pc']==0]
    r1['Remarks'] = "Sulphur is not given"
    hs_data = data1[data1['fuel_type']=='RESIDUAL HS']
    r2 = hs_data[(hs_data['sulphur_content_pc']<0.5) | (hs_data['sulphur_content_pc']>3.5) ]
    r2['Remarks'] = "Fuel Type or Sulphur will be wrong"
    ls_data = data1[(data1['fuel_type']=='RESIDUAL VLS') | (data1['fuel_type']=='DISTILLATE ULS')| (data1['fuel_type']=='DISTILLATE VLS') | (data1['fuel_type']=='RESIDUAL ULS')  | (data1['fuel_type']=='TANKTAINER DISTILLATE VLS')]
    r3 = ls_data[(ls_data['sulphur_content_pc']<0) | (ls_data['sulphur_content_pc']>0.5) ]
    r3['Remarks'] = "Fuel Type or Sulphur will be wrong"
    r4 = data1[data1['sulphur_content_pc'].isnull()]
    r4['Remarks'] = "Sulphur is not given"
    r5 = data1[data1['quantity']<=0]
    r5['Remarks'] = "quantity is not given"
    r6 = data1[data1['quantity'].isnull()]
    r6['Remarks'] = "quantity is not given"
    r7 = data1[data1['bdn_number'].isnull()]
    r7['Remarks'] = "BDN is not given"
    data1["joinall"]= data1['bdn_number'].str.replace(r'[-_\s.,/]+', '')
    data1["is_alpha_Delivery"]= data1["joinall"].str.isalpha()
    r8 = data1[data1['is_alpha_Delivery']==True]
    r8['Remarks'] = "BDN is not given"
    r9 = data1[data1['bdn_number']=='0']
    r9['Remarks'] = "BDN is not given"
    df = pd.concat([r1,r2,r3,r4,r5,r6,r7,r8,r9])
    df.drop('is_alpha_Delivery', axis=1, inplace=True)
    df.drop('joinall', axis=1, inplace=True)
    p = df.groupby('gid')['Remarks'].agg(','.join).reset_index()
    dict_remarks = pd.Series(p['Remarks'].values,index=p['gid']).to_dict()
    import numpy as np
    df['REMARKS'] = np.nan
    df['REMARKS'] = df['REMARKS'].fillna(df['gid'].apply(lambda x:dict_remarks.get(x)))
    df.drop('Remarks', axis=1, inplace=True)
    df.drop_duplicates(keep='first',inplace=True)
    return df
```

```python
#if delivery column starts or ends with underscore(_ and *)
def delivery_modification(FINALNOON):
    FINALNOON["bdn_number"] = FINALNOON["bdn_number"].astype('string')
    FINALNOON_nan=FINALNOON[FINALNOON['bdn_number'].isna()]
    FINALNOON_nan_gid_list = list(FINALNOON_nan['gid'])
    FINALNOON_not_nan=FINALNOON[~FINALNOON['gid'].isin(FINALNOON_nan_gid_list)]
    FINALNOON_not_nan["bdn_number"] = FINALNOON_not_nan["bdn_number"].apply(lambda x: x.strip("_*"))
    FINALNOON_delivery_concat = pd.concat([FINALNOON_not_nan,FINALNOON_nan])
    FINALNOON = FINALNOON.drop(['bdn_number'], axis=1)
    dict_noon_delivery = pd.Series(FINALNOON_delivery_concat['bdn_number'].values,index=FINALNOON_delivery_concat['gid']).to_dict()
    FINALNOON['bdn_number'] = np.nan
    FINALNOON['bdn_number'] = FINALNOON['bdn_number'].fillna(FINALNOON['gid'].apply(lambda x:dict_noon_delivery.get(x)))
    return FINALNOON
```
