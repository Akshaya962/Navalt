```python
import pandas as pd
import json
```

```python
excel_file = "One Atlas.csv"
attributes_json_file = "attributes.json"
other_json_file = "get_all_vdm_vessel_attribute 1.json"
```

```python

# Load Excel data
excel_df = pd.read_csv(excel_file)

# Load JSON data
with open(attributes_json_file, "r") as f:
    attributes = json.load(f)

# Load other JSON file
with open(other_json_file, "r") as f:
    other_data = json.load(f)


# Convert JSON data to DataFrames
attributes_df = pd.DataFrame(attributes)
other_df = pd.DataFrame(other_data)

```

```python

excel_mapping = dict(zip(excel_df["COLUMNS"], excel_df["VALUE"]))
output_data = []
attribute_ids = set()
duplicate_ids = set()

for attribute in attributes:
    attribute_id = attribute["id"]
    attribute_name = attribute["name"]
    
    if attribute_id in attribute_ids:
        duplicate_ids.add(attribute_id)
    else:
        attribute_ids.add(attribute_id)
    
    value = excel_mapping.get(attribute_name, "")
    
    # Construct the output dictionary
    output_data.append({
        "attribute_id": attribute_id,
        "value": value,
        "imo": "9290155",
        "color": "#D32F2F",
        "class": "1",
        "fleet": "1",
        "name": "Atlas",
        "fleet_name": "1",
        "vpq_date": "2025-01-16"
    })

output_data = sorted(output_data, key=lambda x: x["attribute_id"])

if duplicate_ids:
    print(f"Duplicate attribute IDs found: {duplicate_ids}")

with open("output.json", "w") as file:
    json.dump(output_data, file, indent=4)

print("Output JSON saved to 'output.json'")
```
