```python
import pandas as pd
```

```python
data=pd.read_excel("grouped_SPI_combined_voycon.xlsx")
```

```python
data.head()
```

```python
df=data
```

```python
min_rpm = df['RPM'].min()
max_rpm = df['RPM'].max()

print("Minimum RPM:", min_rpm)
print("Maximum RPM:", max_rpm)
```

```python
min_fuel = df['Total_fuel'].min()
max_fuel = df['Total_fuel'].max()

print("Minimum Fuel:", min_fuel)
print("Maximum Fuel:", max_fuel)
```

```python
# Filter rows with RPM less than a threshold (e.g., 1000) and total steaming time equal to 24
filtered_df = df[(df['RPM'] < 100) & (df['TOTAL_STEAMING_TIME'] == 24)]

# Sort the filtered DataFrame by RPM in ascending order and get the top 10 values
top_10_low_rpm = filtered_df.sort_values(by='RPM').head(10)

top_10_low_rpm.head(10)
```
